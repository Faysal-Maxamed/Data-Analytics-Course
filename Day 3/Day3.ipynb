```python
data['Gender'] = data['Gender'].replace("MALE", "Male")
```

```python
data['Gender'] = data['Gender'].replace("male", "Male")
data['Gender'] = data['Gender'].replace("female", "Female")
```

```python
data['City'] = data['City'].replace("Mogadishu", "Mog")
```
