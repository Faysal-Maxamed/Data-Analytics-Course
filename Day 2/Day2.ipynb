# Filtering Data


Numerial Filter

```python
data2[data2['Occupation'] == 'Civil Servant']
```

Filtering multi Categories

```python
data2[data2['Gender'].isin(['Female','NaN'])]
```

# Multi Condition

```python
data2[
    (data2['Gender'] == 'Female') &
    (data2['Occupation'] == 'Civil Servant') &
    (data2['City'] =='Mogadishu')
]
```
